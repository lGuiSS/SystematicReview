```python
import pandas as pd
import io
from urllib.request import urlopen

# 1. Carregar o CSV da Web
url = "https://gist.githubusercontent.com/stevewithington/20a69c0b6d2ff846ea5d35e5fc47f26c/raw/13716ceb2f22b5643ce5e7039643c86a0e0c6da6/country-and-continent-codes-list-csv.csv"
response = urlopen(url)
df_continentes = pd.read_csv(io.StringIO(response.read().decode('utf-8')))

# 2. Criar um dicionário de mapeamento: cca3 -> Continent_Name
# Nota: No CSV, a coluna chama-se 'Three_Letter_Country_Code'
mapeamento_continentes = df_continentes.set_index('Three_Letter_Country_Code')['Continent_Name'].to_dict()

# 3. Atualizar a sua lista de países (paises) que contém a chave 'cca3'
# Supondo que a sua variável se chama 'paises_lista'
for pais in data:
    codigo_iso = pais.get('cca3')
    # Adiciona o nome do continente buscando pelo código cca3
    pais['Continent_Name'] = mapeamento_continentes.get(codigo_iso, "Não Encontrado")

# Exemplo de verificação
print(data[1]) 



output_file = fr"C:\Users\Guilherme\OneDrive\Backup\App RBS\sistema-revisao-bibliografica\src\countryPatterns\countryPatterns_v3.json"
# Salvar o novo arquivo JSON
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
```
